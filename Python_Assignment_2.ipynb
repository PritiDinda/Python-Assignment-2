1. What is a tuple in Python? Explain how tuples are created, accessed, and sliced.
Why are tuples called immutable? Give one real-world situation where a tuple can be useful.

-> A tuple is an ordered collection of elements in Python. It can store multiple values of different data types. Tuples are written using parentheses ()


```python
numbers = (10, 20, 30, 40, 50)
print(numbers)         #creating a tuple

print(numbers[0])
print(numbers[2])
print(numbers[-1])    #Tuple elements are accessed using index numbers, starting from 0

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])    #Slicing a Tuple
```

- Tuples are called immutable because their elements cannot be changed, added, or removed after the tuple is created.

```python
point = (10, 20)
print(point)
#Here, (10, 20) represents the fixed x and y coordinates of a point.
```

2. What is a dictionary? Explain how key-value pairs work in Python. Describe how to add, modify, delete, and access dictionary elements

- A dictionary in Python is one of the most powerful and widely used data structures. It allows you to store data in the form of key-value pairs, making it easy to organize, access, and modify information quickly.

  Dictionaries are created using curly braces {}, with each key-value pair separated by a colon (:) and each pair separated by a comma.

```python
student = {
    "name": "Priti Dinda",
    "age": 22,
    "course": "Data Science"
}

print(student)
```

```python
student["city"] = "Kolkata"  #adding an element
student["age"] = 21          #modifying an element
del student["age"]           #deleting an element
print(student["name"])       #accessing element
```

3. What is a set in Python? Explain why duplicate values are not stored in a set.
Describe union, intersection, and difference with suitable examples.

- A set is an unordered collection of unique values. Sets are created using {}.
- Duplicate values are not stored because a set automatically keeps only unique elements.

```python
numbers = {1, 2, 2, 3}
print(numbers)
```

```python
#Union
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)  #combines elements from both sets.

#Intersection
print(a & b)  #return common elements

#Difference
print(a - b)  #returns elements present in the first set but not the second.
```

4. What are conditional statements in Python? Explain if, if-else, if-elif-else, and nested
if statements with suitable examples

- Conditional statements are used to execute different blocks of code depending on conditions.

```python
#if statements
age = 22

if age >= 18:
    print("Adult")   #Executes code when the condition is true
```

```python
#if-else statements
age = int(input("Enter your age: "))

if age >= 18:
    print("Adult")
else:
    print("Minor")   #Chooses between two blocks.
```

```python
#if-elif-else statements
marks = int(input("Enter your score: "))

if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")      #Checks multiple conditions.
```

```python
age = 20

if age >= 18:
  print('You are an adult')
  if age <= 50:
      print("Eligible")   #An if statement inside another if.
else:
  print('You are under 18')
```

5. Explain the difference between for and while loops. Also explain the
purpose of break, continue, pass, and the else clause with loops.

- A for loop is generally used when we know how many times we want to repeat something.

- A while loop repeats as long as a condition is true.

```python
for i in range(1, 6):
    print(i)

i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
for i in range(5):
    if i == 3:
        break
    print(i)

for i in range(5):
    if i == 3:
        continue
    print(i)

for i in range(5):
    pass

for i in range(3):
    print(i)
else:
    print("Loop completed")
```

6. Create the following tuple:

marks = (78, 85, 90, 67, 88)

Write a Python program to:
* Display the first and last values.
* Display values from index 1 to 3.
* Find the highest and lowest marks.
* Convert the tuple into a list.
* Add 95 to the list.
* Convert the updated list back into a tuple.


```python
marks = (78, 85, 90, 67, 88)

# 1. First and last values
print("First value:", marks[0])
print("Last value:", marks[-1])

# 2. Values from index 1 to 3
print("Index 1 to 3:", marks[1:4])

# 3. Highest and lowest marks
print("Highest:", max(marks))
print("Lowest:", min(marks))

# 4. Convert tuple into list
marks_list = list(marks)

# 5. Add 95
marks_list.append(95)

# 6. Convert list back to tuple
marks = tuple(marks_list)

print("Updated tuple:", marks)
```

7. Create a dictionary to store:
● Student name
● Roll number
● Course
● Marks
Write a Python program to:
- Display all student details.
- Add the student's city.
- Update the marks.
- Display all keys and values.
- Iterate through the dictionary and print each key with its value.

```python
student = {
    "name": "Priti Dinda",
    "roll_no": 101,
    "course": "Data Science",
    "marks": 85
}

# 1. Display all details
print(student)

# 2. Add city
student["city"] = "Kolkata"

# 3. Update marks
student["marks"] = 90

# 4. Display all keys and values
print(student.keys())
print(student.values())

# 5. Print each key with its value
for key, value in student.items():
    print(key, ":", value)
```

8. Two classes have registered for different activities.
sports = {"Rahul", "Anil", "Priya", "Kiran"}
music = {"Priya", "Kiran", "Meena", "Arun"}
Write a Python program to find:
- Students who joined either activity.
- Students who joined both activities.
- Students who joined only sports.
- Students who joined only music.


```python
sports = {"Rahul", "Anil", "Priya", "Kiran"}
music = {"Priya", "Kiran", "Meena", "Arun"}

# 1. Either activity
print("Either activity:", sports | music)

# 2. Both activities
print("Both activities:", sports & music)

# 3. Only sports
print("Only sports:", sports - music)

# 4. Only music
print("Only music:", music - sports)
```

9.  Consider the following string:
"Python Java C C++ Python Java Python"

Write a Python program to:
- Convert the string into a list of programming languages.
- Display the list.
- Count how many times "Python" appears.
- Remove duplicate language names.
- Join the unique language names using " | ".


```python
text = "Python Java C C++ Python Java Python"

# 1. Convert string into list
languages = text.split()

# 2. Display list
print("List:", languages)

# 3. Count Python
print("Python count:", languages.count("Python"))

# 4. Remove duplicates
unique_languages = list(dict.fromkeys(languages))
print("Unique languages:", unique_languages)

# 5. Join unique languages
result = " | ".join(unique_languages)
print("Joined:", result)
```

10. Write a Python program to print multiplication tables from 1 to 5.
Use nested loops to display each table from 1 × 1 up to 5 × 10.
Also modify the program so that:
● The number 5 is skipped using continue.
● The inner loop stops when the multiplier reaches 8 using break


```python
for i in range(1, 6):
    print("Table of", i)

    for j in range(1, 11):
        print(i, "x", j, "=", i * j)

    print()
```

```python
for i in range(1, 6):
    if i == 5:
        continue

    print("Table of", i)

    for j in range(1, 11):
        if j == 8:
            break

        print(i, "x", j, "=", i * j)

    print()
```
